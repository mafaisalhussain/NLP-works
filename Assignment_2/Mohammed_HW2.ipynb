NLP Assignment: Text Processing Pipeline
CNN Article: "Why Luxury Brands Are So Expensive"
URL: https://www.cnn.com/2025/06/13/style/why-luxury-brands-are-so-expensive

Steps:
  1. Read the file / fetch article text
  2. Remove HTML tags
  3. Lowercase and remove punctuation
  4. Remove stopwords (NLTK)
  5. Lemmatize and compare with stemming

```python
#Read the File

import requests

url = "https://www.cnn.com/2025/06/13/style/why-luxury-brands-are-so-expensive"

headers = {
    "User-Agent": "Mozilla/5.0"
}

response = requests.get(url, headers=headers)
html_content = response.text

print("FIRST 700 CHARACTERS OF DOWNLOADED FILE:\n")
print(html_content[:700])
```

```python
from bs4 import BeautifulSoup
import json

soup = BeautifulSoup(html_content, "lxml")

article_text = ""

# find all json-ld scripts
scripts = soup.find_all("script", type="application/ld+json")

for script in scripts:
    try:
        data = json.loads(script.string)

        # Sometimes it's a list
        if isinstance(data, list):
            for item in data:
                if item.get("@type") == "NewsArticle":
                    article_text += item.get("articleBody", "")
        else:
            if data.get("@type") == "NewsArticle":
                article_text += data.get("articleBody", "")

    except:
        pass

print("FIRST 700 CHARACTERS OF ARTICLE TEXT:\n")
print(article_text[:700])
```

```python
#Remove HTML Tags

clean_text = BeautifulSoup(article_text, "html.parser").get_text()

print(clean_text[:700])
```

```python
#Lowercase & Remove Punctuation

import string

text_lower = clean_text.lower()
text_no_punct = text_lower.translate(str.maketrans('', '', string.punctuation))

print(text_no_punct[:700])
```

```python
import nltk

nltk.download('punkt')
nltk.download('punkt_tab')
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('omw-1.4')
```

```python
#Remove Stopwords

from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

tokens = word_tokenize(text_no_punct)

stop_words = set(stopwords.words('english'))

filtered_words = [w for w in tokens if w not in stop_words and w.isalpha()]

print(filtered_words[:100])
```

```python
#Lemmatization

from nltk.stem import WordNetLemmatizer

lemmatizer = WordNetLemmatizer()

lemmatized_words = [lemmatizer.lemmatize(word) for word in filtered_words]

print("FIRST 50 LEMMATIZED WORDS:\n")
print(lemmatized_words[:50])
```

```python
#stemming
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

stemmed_words = [stemmer.stem(word) for word in filtered_words]

print("FIRST 50 STEMMED WORDS:\n")
print(stemmed_words[:50])
```
