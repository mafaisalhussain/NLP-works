```python
# Step 1: Load a fairy tale text as training data

import requests

# URL of Grimm's Fairy Tales from Project Gutenberg
url = "https://www.gutenberg.org/files/2591/2591-0.txt"

response = requests.get(url)
text = response.text

# Display first 1000 characters to confirm it loaded
print(text[:5000])
```

```python
# Step 2: Preprocess the text and build the Markov chain

import re
from collections import defaultdict

# Clean the text
clean_text = re.sub(r'[^a-zA-Z\s]', '', text)  # remove punctuation
clean_text = clean_text.lower()

# Split into words
words = clean_text.split()

# Build the Markov chain dictionary
markov_chain = defaultdict(list)

for i in range(len(words) - 1):
    current_word = words[i]
    next_word = words[i + 1]
    markov_chain[current_word].append(next_word)

# Show a few transitions
for key in list(markov_chain.keys())[:10]:
    print(key, "->", markov_chain[key][:5])
```

```python
# Step 3: Generate a story starting with the required phrase

import random

# Starting phrase required by the assignment
start_phrase = "Once upon a time there was a kingdom"

# Convert start phrase to lowercase words to match training data
story_words = start_phrase.lower().split()

# Number of words to generate
num_words = 200

current_word = story_words[-1]

for _ in range(num_words):

    if current_word in markov_chain:
        next_word = random.choice(markov_chain[current_word])
        story_words.append(next_word)
        current_word = next_word
    else:
        break

# Join the words into a story
story = " ".join(story_words)

# Capitalize first letter
story = story.capitalize()

print(story)
```

```python
# Step 2: Load a text generation pipeline

from transformers import pipeline

# Load GPT-2 model for text generation
generator = pipeline("text-generation", model="gpt2")

print("Model loaded successfully!")
```

```python
# Step 3: Generate a creative story about why studying data science is important today

prompt = """
Write a creative short story about why studying data science today is important.
The story should involve a student who discovers the power of data to solve real-world problems.
"""

story = generator(
    prompt,
    max_length=200,
    num_return_sequences=1,
    temperature=0.9
)

print(story[0]['generated_text'])
```
