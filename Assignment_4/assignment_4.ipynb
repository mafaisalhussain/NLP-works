```python
import spacy
import benepar

# Load spaCy
nlp = spacy.load("en_core_web_sm")

# Add benepar parser
nlp.add_pipe("benepar", config={"model": "benepar_en3"})

# Gettysburg Address sentences
sentence1 = """Four score and seven years ago our fathers brought forth on this continent, a new nation, conceived in Liberty, and dedicated to the proposition that all men are created equal."""

sentence2 = """Now we are engaged in a great civil war, testing whether that nation, or any nation so conceived and so dedicated, can long endure."""

doc1 = nlp(sentence1)
doc2 = nlp(sentence2)

# -------- spaCy tokenization --------
print("spaCy Tokens — Sentence 1:")
print([token.text for token in doc1])

print("\nspaCy Tokens — Sentence 2:")
print([token.text for token in doc2])

# -------- benepar tokenization --------
print("\nbenepar Tokens — Sentence 1:")
sent1 = list(doc1.sents)[0]
print([token.text for token in sent1])

print("\nbenepar Tokens — Sentence 2:")
sent2 = list(doc2.sents)[0]
print([token.text for token in sent2])
```

```python
import benepar
import nltk
from nltk.tree import Tree

# Load benepar parser (independent of spaCy)
parser = benepar.Parser("benepar_en3")

sentence1 = """Four score and seven years ago our fathers brought forth on this continent, a new nation, conceived in Liberty, and dedicated to the proposition that all men are created equal."""

# Tokenize sentence using NLTK
tokens = nltk.word_tokenize(sentence1)

# Parse sentence
tree = parser.parse(tokens)

# Print constituency tree
print(tree)
```

5. Extract Noun Phrases
Using spaCy, extract all noun phrases (noun chunks) from both sentences.

```python
import spacy

# Load spaCy model
nlp = spacy.load("en_core_web_sm")

# Gettysburg Address sentences
sentence1 = """Four score and seven years ago our fathers brought forth on this continent, a new nation, conceived in Liberty, and dedicated to the proposition that all men are created equal."""

sentence2 = """Now we are engaged in a great civil war, testing whether that nation, or any nation so conceived and so dedicated, can long endure."""

# Process sentences
doc1 = nlp(sentence1)
doc2 = nlp(sentence2)

# -------- Sentence 1 --------
print("Noun Phrases in Sentence 1:\n")
for chunk in doc1.noun_chunks:
    print(chunk.text)

# -------- Sentence 2 --------
print("\nNoun Phrases in Sentence 2:\n")
for chunk in doc2.noun_chunks:
    print(chunk.text)
```

6. CRF and HMM
Why do you use CRF and HMM? How do they differ? Please summarize in less than 50
words.

HMM and CRF are sequence labeling models used for tasks like POS tagging and NER.
HMM is generative and assumes independent observations.
CRF is discriminative, considers full context and overlapping features, and usually achieves higher accuracy because it models conditional probabilities directly without strong independence assumptions.
